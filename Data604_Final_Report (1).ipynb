# Introduction

Alberta's economny is substantially dependent on its Oil and Energy sector.The motivation behind this project is to explore the relationship between stock prices of these oil companies and various macroeconomic indicators to gain insights into how external economic factors influence market movements. Stock prices are affected by a multitude of factors, including oil prices, inflation rates, GDP growth, and geopolitical events. Understanding these relationships is crucial for investors, policymakers, and financial analysts who seek to interpret market trends.
Our key research questions include:

*  How do macroeconomic indicators correlate with stock price movements?

*  Which macroeconomic factors have the strongest influence on stock volatility?

*  How do different companies in the energy sector react to external economic changes?

We chose this project because of its relevance to financial decision-making and risk assessment in the oil and gas industry, which plays a crucial role in the Canadian economy.
